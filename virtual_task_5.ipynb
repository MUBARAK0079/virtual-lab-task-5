```python
import pandas as pd
import plotly.express as px
from plotly.subplots import make_subplots
import plotly.graph_objects as go
```

```python
import pandas as pd
import plotly.express as px

df = px.data.gapminder()

df.head()
```

```python
print(df.shape)

df.info()

df.describe()

df.isnull().sum()
```

```python
# Convert Order Date
# df['Order Date'] = pd.to_datetime(df['Order Date']) # This column does not exist in the current DataFrame.

# Extract Year and Month
# df['Year'] = df['Order Date'].dt.year # 'Order Date' column not found.
# df['Month'] = df['Order Date'].dt.strftime('%b') # 'Order Date' column not found.

df.head()
```

```python
category_sales = df.groupby('continent')['gdpPercap'].sum().reset_index()

fig = px.bar(category_sales,
             x='continent',
             y='gdpPercap',
             color='continent',
             title='Total GDP per Capita by Continent')

fig.show()
```

```python
region_profit = df.groupby('continent')['pop'].sum().reset_index()

fig = px.pie(region_profit,
             names='continent',
             values='pop',
             title='Population Distribution by Continent')

fig.show()
```

```python
yearly_population = df.groupby('year')['pop'].sum().reset_index()

fig = px.line(yearly_population,
              x='year',
              y='pop',
              markers=True,
              title='Population Trend Over Years')

fig.show()
```

```python
continent_life_exp = df.groupby('continent')['lifeExp'].mean().reset_index()

fig = px.bar(continent_life_exp,
             x='continent',
             y='lifeExp',
             color='continent',
             title='Average Life Expectancy by Continent')

fig.show()
```

```python
top_countries_pop = df.groupby('country')['pop'].max().nlargest(10).reset_index()

fig = px.bar(top_countries_pop,
             x='pop',
             y='country',
             orientation='h',
             color='pop',
             title='Top 10 Countries by Population (Latest Year)')

fig.show()
```

```python
fig = px.scatter(df,
                 x='gdpPercap',
                 y='lifeExp',
                 color='continent',
                 size='pop',
                 hover_name='country',
                 hover_data=['year'],
                 title='Life Expectancy vs. GDP per Capita')

fig.show()
```

```python
latest_year_df = df[df['year'] == df['year'].max()]

fig = px.choropleth(
    latest_year_df,
    locations='iso_alpha',
    color='gdpPercap',
    hover_name='country',
    color_continuous_scale=px.colors.sequential.Plasma,
    title='GDP per Capita by Country (Latest Year)'
)

fig.show()
```

```python
print('Total Population (Latest Year):', round(df.groupby('country')['pop'].max().sum(), 2))
print('Average Life Expectancy (Overall):', round(df['lifeExp'].mean(), 2))
print('Average GDP per Capita (Overall):', round(df['gdpPercap'].mean(), 2))
print('Number of Unique Countries:', df['country'].nunique())
```

```python
fig = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=(
        "Population by Continent",
        "Life Expectancy by Continent",
        "GDP per Capita Trend Over Years",
        "Life Expectancy vs. GDP per Capita"
    ),
    specs=[[{"type":"bar"},{"type":"pie"}],
           [{"type":"scatter"},{"type":"scatter"}]]
)

# Population by Continent
continent_pop = df.groupby('continent')['pop'].sum().reset_index()
fig.add_trace(
    go.Bar(
        x=continent_pop['continent'],
        y=continent_pop['pop'],
        name='Population',
        marker_color=px.colors.qualitative.Plotly # Using Plotly's default qualitative colors
    ),
    row=1,
    col=1
)

# Life Expectancy by Continent
continent_life = df.groupby('continent')['lifeExp'].mean().reset_index()
fig.add_trace(
    go.Pie(
        labels=continent_life['continent'],
        values=continent_life['lifeExp'],
        name='LifeExp'
    ),
    row=1,
    col=2
)

# GDP per Capita Trend Over Years
year_gdp = df.groupby('year')['gdpPercap'].mean().reset_index()
fig.add_trace(
    go.Scatter(
        x=year_gdp['year'],
        y=year_gdp['gdpPercap'],
        mode='lines+markers',
        name='GDP per Capita',
        line=dict(color='orange')
    ),
    row=2,
    col=1
)

# Life Expectancy vs. GDP per Capita (colored by Year)
fig.add_trace(
    go.Scatter(
        x=df['gdpPercap'],
        y=df['lifeExp'],
        mode='markers',
        marker=dict(
            color=df['year'],
            colorscale='Viridis',
            showscale=True, # Show color scale for year
            size=df['pop']/1e6, # Scale pop for marker size
            sizemode='area',
            sizeref=2.*df['pop'].max()/(40.**2) # Adjust size reference for better visualization
        ),
        name='LifeExp vs GDP',
        hovertext=df['country'] + ' (' + df['year'].astype(str) + ')' # Add country and year to hover
    ),
    row=2,
    col=2
)

fig.update_layout(
    height=800,
    width=1200, # Increased width for better readability
    title='Interactive Gapminder Dashboard',
    showlegend=False # Hide individual trace legends, subplot titles are sufficient
)

fig.show()
```
